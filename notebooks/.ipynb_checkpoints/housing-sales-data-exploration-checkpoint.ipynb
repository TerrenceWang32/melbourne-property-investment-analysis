Upon the initial exploration of the data collected, different structures of data were identified.
Thus functions with certain similarities, while also having diferences to clean the different groups of data
need to be created separately depending on the structure of the data.

```python
## importing the packages used for the project
import pandas as pd
from pathlib import Path
```

```python
## function to clean the data without indication of the level of sales, ^ or *

def df_creation(file_path, colnames):
    df = pd.read_excel(file_path, sheet_name='Sheet1', header=None)
    df = df.drop(columns=[6,7,8,9]) # based on the file structure, 6-9 columns will not be used
    df.columns = colnames
    # pattern was found that ABBOTSFORD is always the first record, so remove anything above it
    matches = df.index[df['suburb'] == 'ABBOTSFORD']
    if len(matches) > 0:
        df_cleaned = df.loc[matches[0]:].reset_index(drop=True)
    # removing the last line explaining the meaning of ^ and * symbols
    df_cleaned = df_cleaned.drop(df_cleaned.index[-1])
    # changing the type of data into numerical data for future analysis
    df_cleaned[colnames[1:]] = df_cleaned[colnames[1:]].apply(pd.to_numeric, errors='coerce').astype('Int64')
    # longing the dataframe to make it about suburb and their quarterly performance
    value_cols = df_cleaned.columns[1:6].tolist()
    df_melted = pd.melt(df_cleaned, id_vars=['suburb'], value_vars = value_cols)
    mel_cols = ['suburb','quarter','median_price']
    df_melted.columns = mel_cols
    df_melted = df_melted.sort_values(['suburb','quarter']).reset_index(drop=True)
    # adding the source_file column to the dataframe for future analysis
    df_melted['source_file'] = Path(file_path).name
    return df_melted
```

```python
# cleaning and creating dataframe for Q4 2023
p1 = "../data/raw/Median-House-Q4-2023.xls"
c1 = ['suburb','2022Q4','2023Q1','2023Q2','2023Q3','2023Q4']
sales_2023 = df_creation(p1,c1)
sales_2023.head(20)
```

```python
# cleaning and creating dataframe for Q4 2024
p = "../data/raw/median-house-q4-2024.xls"
c = ['suburb','2023Q4','2024Q1','2024Q2','2024Q3','2024Q4']
sales_2024 = df_creation(p,c)
sales_2024.head(20)
```
